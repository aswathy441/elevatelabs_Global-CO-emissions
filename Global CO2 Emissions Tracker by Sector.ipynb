```python
df.isnull().sum()
```

```python
object_cols = df.select_dtypes(include=['object']).columns.tolist()

for col in object_cols:
    df[col] = df[col].fillna(df[col].mode()[0])

df
```

```python
if 'Unnamed: 0' in df.columns:
    df.drop('Unnamed: 0', axis=1, inplace=True)

df.head()
```

```python
df.isnull().sum()
```

```python
#Prepare per capita and per GDP metrics
df['co2_per_capita'] = df['co2'] / df['population']
df['co2_per_gdp'] = df['co2'] / (df['gdp'] / 1e6)  # divide GDP by 1M for scale
df
```

```python
#Sector-wise per-capita metrics
sectors = ['cement_co2','coal_co2','oil_co2','gas_co2']
for s in sectors:
    df[f'{s}_per_capita'] = df[s] / df['population']
df
```

```python
df.to_csv("cleaned_co2_data.csv", index=False)
```
